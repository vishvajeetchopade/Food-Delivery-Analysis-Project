```python
import pandas as pd
import numpy as np

```

```python
df = pd.read_csv("Food_Delivery .csv")
```

```python
df.head()
```

```python
print(df.shape)
df.info()
print(df.describe(include="all").T)
print(df.isnull().sum())
print("Duplicates:", df.duplicated().sum())
for c in ["city", "restaurant", "item", "payment_method"]:
    print(df[c].value_counts(dropna=False))
print((df.delivery_time_minutes <= 0).sum(), (df.delivery_time_minutes > 120).sum(), (df.quantity > 10).sum())
```

```python
df["order_date"] = pd.to_datetime(df["order_date"])
df["city"] = df["city"].str.strip().str.title().fillna("Unknown")

df.loc[(df.delivery_time_minutes <= 0) | (df.delivery_time_minutes > 120), "delivery_time_minutes"] = np.nan
df["delivery_time_minutes"] = df.groupby("restaurant")["delivery_time_minutes"] \
                                .transform(lambda s: s.fillna(s.median()))

df.loc[df.quantity > 4, "quantity"] = 4
```

```python
df = df.drop_duplicates()
df = df.drop_duplicates(subset="order_id", keep="first").reset_index(drop=True)

rated = df.dropna(subset=["rating"])
df["order_value"] = df.quantity * df.price * (1 - df.discount / 100)  # discount treated as %
df["hour"] = df.order_date.dt.hour
df["day_name"] = df.order_date.dt.day_name()
df["month"] = df.order_date.dt.to_period("M")
```

```python
df.delivery_time_minutes.describe()
df.groupby("restaurant").delivery_time_minutes.agg(["mean", "median", "std"]).sort_values("mean")
df.groupby("city").delivery_time_minutes.mean()
df.groupby("hour").delivery_time_minutes.mean()
df["delivery_bucket"] = pd.cut(df.delivery_time_minutes, [0,20,30,45,60,120],
                               labels=["<=20","21-30","31-45","46-60","60+"])
df.delivery_bucket.value_counts(normalize=True).sort_index()
df[["delivery_time_minutes", "rating"]].corr()
```

```python
df.payment_method.value_counts(normalize=True)
df.groupby("payment_method").agg(orders=("order_id","count"), revenue=("order_value","sum"),
                                 avg_order_value=("order_value","mean"), avg_rating=("rating","mean"))
pd.crosstab(df.city, df.payment_method, normalize="index")
df.groupby(["month","payment_method"]).size().unstack()
```

```python
cust = df.groupby("customer_id").agg(orders=("order_id","count"), total_spent=("order_value","sum"),
                                     avg_order=("order_value","mean"), avg_rating=("rating","mean"))
cust.sort_values("total_spent", ascending=False).head(10)
(cust.orders > 1).mean()                                   # repeat-customer rate
df.groupby("customer_id").restaurant.nunique().value_counts()
df.groupby("customer_id").item.agg(lambda s: s.mode()[0]).value_counts()
df["discount_band"] = pd.cut(df.discount, [-1,10,20,30], labels=["0-10%","11-20%","21-30%"])
df.groupby("discount_band", observed=True).agg(avg_qty=("quantity","mean"), avg_value=("order_value","mean"))
```

```python
df.groupby("month").agg(orders=("order_id","count"), revenue=("order_value","sum"))
df.groupby(df.order_date.dt.to_period("W")).size()
df.groupby("hour").size()
df.groupby("day_name").size()
df.groupby("item").agg(orders=("order_id","count"), revenue=("order_value","sum")).sort_values("revenue", ascending=False)
df.groupby("city").agg(orders=("order_id","count"), revenue=("order_value","sum"))
```

```python
rest = df.groupby("restaurant").agg(orders=("order_id","count"), revenue=("order_value","sum"),
                                    avg_delivery=("delivery_time_minutes","mean"),
                                    avg_rating=("rating","mean"), avg_discount=("discount","mean"))
print(rest.sort_values("revenue", ascending=False))
print("Total revenue:", df.order_value.sum(), "| AOV:", df.order_value.mean())
print("Discount cost:", (df.quantity * df.price * df.discount / 100).sum())
print("Late deliveries (>45 min):", (df.delivery_time_minutes > 45).mean())
```
